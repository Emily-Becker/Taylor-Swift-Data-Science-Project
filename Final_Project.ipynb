```python
danceability = taylor_data["danceability"]
energy = taylor_data["energy"]

plt.scatter(danceability, energy, s=15, color="lightseagreen", marker="+")
plt.xlabel("Danceability", fontsize=14)
plt.ylabel("Energy", fontsize=14)
```

```python
corr = scipy.stats.pearsonr(danceability, energy)
print(corr)

linear_model = scipy.stats.linregress(danceability, energy)
print(linear_model)

slope = linear_model[0]
y_intercept = linear_model.intercept
linear_fit = slope*danceability+y_intercept

plt.scatter(danceability, energy, s=15, color="lightseagreen", marker="+")
plt.plot(danceability, linear_fit, color="red", linewidth = 3)
plt.xlabel("Danceability", fontsize=14)
plt.ylabel("Energy", fontsize=14)
```

```python
data = {
    'album': ['1989', '1989 (Deluxe Edition)', 'reputation', 'Lover', 'folklore', 'folklore (deluxe version)', 'folklore: the long pond studio sessions', 'evermore', "evermore (deluxe version)", "Fearless (Taylor's Version)", "Red (Taylor's Version)", 'Midnights', 'Midnights (3am Edition)', 'Midnights (The Til Dawn Edition)', "Speak Now (Taylor's Version)", "1989 (Taylor's Version)", "1989 (Taylor's Version) [Deluxe]", 'THE TORTURED POETS DEPARTMENT','THE TORTURED POETS DEPARTMENT:THE ANTHOLOGY'],
    'release_date': ['2014-10-27', '2014-10-27', '2017-11-10', '2019-08-23', '2020-07-24', '2020-08-18', '2020-11-25', '2020-12-11', '2021-01-07', '2021-04-09', '2021-11-12', '2022-10-21', '2022-10-22', '2023-05-26', '2023-07-07', '2023-10-26', '2023-10-27', '2024-04-18', '2024-04-19'],
}
df = pd.DataFrame(data)
df['release_date'] = pd.to_datetime(df['release_date'])
```

```python
import matplotlib.cm as cm
# Sort by date
df = df.sort_values('release_date')

# Create a color palette
colors = cm.get_cmap('tab20', len(df))

# Plot
plt.figure(figsize=(12, 6))

for i, row in df.iterrows():
    color = colors(i)
    plt.hlines(y=row['album'], xmin=row['release_date'], xmax=row['release_date'] + pd.to_timedelta(10, unit='d'),
               color=color, linewidth=5)
    plt.scatter(row['release_date'], row['album'], color=color, s=100, edgecolor='black', zorder=3)

plt.title("Taylor Swift Album Release Timeline", fontsize=14, fontweight='bold')
plt.xlabel("Release Date")
plt.tight_layout()
plt.grid(True, axis='x', linestyle='--', alpha=0.5)
plt.show()
```
