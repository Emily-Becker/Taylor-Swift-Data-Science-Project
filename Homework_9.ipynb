```python
# I want to find out if there is a connection between explicitness and the different traits ranked at the end
# (danceability-valence). I want to do this because often times, Taylor Swift fans will notice that different tracks from her
# albums will have different vibes or feels. For example, Track 5 songs are said to be more emotional and heartbreaking, whereas
# Tracks 1-3 are normally singles and are more upbeat and happy or energetic. I will be able to evaluate this using machine learning and KNN.

# This fits the description from the cell above because the explicit column will only need to be chang the data type to category prior to using it.
# This will not need to be classified into different "classes" or "ranges". Because I am simply converting non-numeric data to numeric data, this is
# already formatted the way it needs to be and it will be ready for classification after changing this to numbers.
```

```python
# Pandas import
import pandas as pd
# Matplotlib import
import matplotlib.pyplot as plt
# Sci-kit learn imports
#k-Nearest Neighbors
from sklearn.neighbors import KNeighborsClassifier
# Train-test split
from sklearn.model_selection import train_test_split
# Accuracy Score
from sklearn.metrics import accuracy_score
# Confusion Matrix
from sklearn.metrics import ConfusionMatrixDisplay
```

```python
# To start, I needed to change the data type of explicit from boolean to an integer
# This is done below so that False=0 and True=1
taylor_data["explicit"] = taylor_data["explicit"].astype(int)
```

```python
taylor_data # Shows new data where False=0 and True=1
taylor_data.describe() # Shown here to show that explicit values can be used as integers now
```

```python
# Can now start the machine learning workflow below
# Splitting the data into X and y data
X = taylor_data[["danceability","energy","loudness","speechiness","acousticness","instrumentalness","liveness","valence"]]
y = taylor_data["explicit"]

# Perform a train test split on the data wtih 20% of the data being randomly
# set aside for the test set
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20)
```

```python
# Define the classifier with the given number of neighbors
classifier = KNeighborsClassifier(n_neighbors=5)
# Fit the model with the training data
classifier.fit(X_train, y_train)
# Create the predictions
predictions = classifier.predict(X_test)
# Find the accuracy score
accuracy = accuracy_score(y_test, predictions)
print(accuracy*100)
ConfusionMatrixDisplay.from_estimator(classifier, X_test, y_test)
```

```python
# This confusion matrix shows that based on those values (danceability-valence) this model can predict some of these values accurately.
# 66 were predicted to not be explicit and were not explicit
# 12 were predicted to not be explicit but were not actually explicit
# 6 were predicted to be explicit but were actually explicit
# 3 were predicted to be explicit and were explicit

# This also shows a decent accuracy score based off of these variables and 5 nearest neighbors with it being 79.31%
# This can be improved using the code below (hyperparameter tuning)
```

```python
# Using empty lists to fill and using a for loop to do that
k_values = []
accuracies = []
for k in range(1, 15, 1):
  knn = KNeighborsClassifier(n_neighbors=k)
  knn.fit(X_train, y_train)
  predictions = knn.predict(X_test)
  accuracy = accuracy_score(predictions, y_test)
  k_values.append(k)
  accuracies.append(accuracy)
# Plotting the values created using this for loop
plt.plot(k_values, accuracies)
plt.xlabel("Number of Neighbors")
plt.ylabel("Accuracy")
plt.grid()
```

```python
# According to this, 2 nearest neighbors would make the model more accuarate (note that I did not use 1 because that I think it should
# be multiple neighbors, not just a singular one). Below I reran the model using 2 nearest neighbors

knn = KNeighborsClassifier(n_neighbors=2)
knn.fit(X_train, y_train)
predictions = knn.predict(X_test)
accuracy = accuracy_score(predictions, y_test)
print("Accuracy:", accuracy*100)
ConfusionMatrixDisplay.from_estimator(knn, X_test, y_test)
```

```python
# This resulted in an improved confusion matrix and a better accuracy score of 85.06% rather than 79.31%
# This shows that the model has been improved and there is about an 85.06% chance that I can predict if a track is explicit
# or not by its traits listed in this data set using its 2 nearest neighbors.
```

```python
# For this problem, I want to see if duration/runtime is can be predicted when you know the explicitness and track number. I want to
# do this because it is interesting to me to see if different variables like this really impact how long a song may be.
# I can do this using machine learning.

# This task is fulfilling the requirements of this problem because I will need to classify the duration of the songs before starting. This means
# that I will need to create different numeric values to represent the different lengths of songs (short, medium, and long length) before using
# these values since there are some many possibilities of what this could be in this data.
```

```python
#Using the describe function to look at the highest and lowest duration time to determine a short, medium, and long length
taylor_data.describe()
```

```python
# I will start by classifying the duration column of data using the function below

def duration_category(duration):
    """
    Converts the duration of the song in ms to a cateogorical variable,
    where 0 represents a short song and 1 represents a medium length song,
    and 2 represents a long song.
    Input:
        duration (float): the duration of the song in ms
    Returns:
        0/1/2 (int): the label for the categorical variable
    """
    if duration < 209820.00000: # Short length songs (number based on the 25% above)
        return 0
    elif duration < 255579.750000: # Medium length songs (number based on the 75% above)
        return 1
    else:
        return 2 # Long length songs
```

```python
# The map function will apply the given function to every entry in the alcohol column
# I will save the results as a new column in the dataframe
taylor_data["duration_lengths"] = taylor_data["duration_ms"].map(duration_category)
```

```python
taylor_data.info() # Shows the new column as integers
taylor_data.nunique() # Shows the 3 unique values in this column
# This shows me that I can start the machine learning workflow
```

```python
# Splitting the data into X and y data
# Note that I am using the explicit values from problem 1
X_p2 = taylor_data[["explicit","track_number"]]
y_p2 = taylor_data["duration_lengths"]

# Perform a train test split on the data wtih 20% of the data being randomly
# set aside for the test set
X_p2_train, X_p2_test, y_p2_train, y_p2_test = train_test_split(X_p2, y_p2, test_size=0.20)
```

```python
# Define the classifier with the given number of neighbors
classifier = KNeighborsClassifier(n_neighbors=5)
# Fit the model with the training data
classifier.fit(X_p2_train, y_p2_train)
# Create the predictions
predictions = classifier.predict(X_p2_test)
# Find the accuracy score
accuracy = accuracy_score(y_p2_test, predictions)
print(accuracy*100)
ConfusionMatrixDisplay.from_estimator(classifier, X_p2_test, y_p2_test)
```

```python
# This confusion matrix shows that based on the X data, this model can predict some of the y-values accurately, but not the majority.
# 6 were predicted to short songs and were short songs
# 28 were predicted to be medium-length songs and were medium-length songs
# 6 were predicted to be long songs and were long songs
# The rest of the data was inaccurately predicted, meaning that this algorithm is not the most effective
# In fact, the accuracy score was only 45.98%

# This can be used using hyperparameter tuning in the cells below.
```

```python
# Using empty lists to fill and using a for loop to do that
k_p2_values = []
p2_accuracies = []
for k in range(1, 15, 1):
  knn_p2 = KNeighborsClassifier(n_neighbors=k)
  knn_p2.fit(X_p2_train, y_p2_train)
  predictions = knn_p2.predict(X_p2_test)
  accuracy = accuracy_score(predictions, y_p2_test)
  k_p2_values.append(k)
  p2_accuracies.append(accuracy)
# Plotting the values created using this for loop
plt.plot(k_p2_values, p2_accuracies)
plt.xlabel("Number of Neighbors")
plt.ylabel("Accuracy")
plt.grid()
```

```python
# According to this, 13 nearest neighbors would make the model more accuarate.
# Below I reran the model using 13 nearest neighbors

knn_p2 = KNeighborsClassifier(n_neighbors=13)
knn_p2.fit(X_p2_train, y_p2_train)
predictions = knn_p2.predict(X_p2_test)
accuracy = accuracy_score(predictions, y_p2_test)
print("Accuracy:", accuracy*100)
ConfusionMatrixDisplay.from_estimator(knn_p2, X_p2_test, y_p2_test)
```

```python
# This confusion matrix shows an improvement in the model and shows an increased accuracy score (increasing from 45.98% to 49.43%)
# While changing the k nearest neighbors to 10 did improve this algorithm, it is still not the most accuarate, as less than half the
# data can be predicted accurately. Because of this, this model is not very accurate, but using hyperparameter tuning, I was able to
# make it a little better.
```
