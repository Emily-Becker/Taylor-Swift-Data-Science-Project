## Question 1

```python
# The first question I would like to answer: Which album has the longest runtime in minutes?
# To do this, I first created a sub-dataframe for each album. Then, I summed up the song times using the df["column"].sum()
# Then, I convert those to minutes using a for loop including the album names mask and their times converted to minutes.
```

```python
taylor_data["album_name"].unique() # Used to see all the album names
```

```python
album_names = taylor_data["album_name"].unique() # Used to extract distinct values present within the data like above
times = [] # Used below to save things into this list later on after the for loop
for album in album_names: #For loop using the albums in my data and this will show their names and sum their runtime
  current_album = taylor_data[taylor_data["album_name"]==album]
  runtime = current_album["duration_ms"].sum()
  times.append(runtime/60000) # This is then divided by 60000 to convert to minutes

# Wanted to change the name of the folklore: long pond studios album, but the cell would not let me when reopening Google Colab, apologies for that!
plt.bar(album_names, times) # This graph is then plotted and modified using matplotlib
plt.xlabel("Album Name", fontsize=14)
plt.ylabel("Total Minutes", fontsize=14)
plt.xticks(rotation=90)
```

```python
# This graph shows that the reputation stadium tour playlist has the longest runtime. This is interesting to me because
# I thought the longest runtime would be found in a Deluxe or Taylor's version albums, since they normally have more tracks and
# longer versions of tracks. As it turns out, this is the longest album here, which makes sense as to why it has the longest runtime.
# Also the deluxe albums and some Taylor's version albums tend to have long total time, which makes sense for this data.
```

## Chi Squared Analysis

```python
# The chi squared analysis I decided to look at is similar to what I looked at in the last question. This time I will be answering the
# following: "Is there a correlation between explicitness of tracks and the album they're found on?"

# To do this, I used the data from above similarly and used Pandas to make a cross tab and the scipy stats package to perform the chi
# squared analysis. I then printed the "p" variable from that to look at their correlation, as this value will tell me about that.
```

```python
crosstab = pd.crosstab(taylor_data["explicit"], taylor_data["album_name"])
crosstab
```

```python
crosstab = pd.crosstab(taylor_data["explicit"], taylor_data["album_name"])
c , p , dof, expected = scipy.stats.chi2_contingency(crosstab)
```

```python
print(p)
```

```python
# This shows that the p value is equal to 3.866e-12.
# p values < 0.05 show data that are related/correlated. In this case, p is much smaller than that (due to the larger negative exponent),
# therefore, the explicitness of songs are related to the album they are one. In other words, explicitness is dependent on the album.
```
